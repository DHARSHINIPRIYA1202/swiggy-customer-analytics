TOTAL ORDERS

```python
total_orders = len(orders)

print("Total Orders:", total_orders)
```

TOTAL SALES

```python
total_sales = orders["sales_amount"].sum()

print("Total Sales:", total_sales)
```

AVERAGE ORDER VALUE

TOTAL QUANTITY SOLD

```python
total_quantity = orders["sales_qty"].sum()

print("Total Quantity Sold:", total_quantity)
```

ORDERS OVER TIME

```python
orders["order_date"] = pd.to_datetime(orders["order_date"])
```

```python
daily_orders = (orders.groupby("order_date").size())
```

```python
plt.figure(figsize=(12, 5))

plt.plot(daily_orders.index,daily_orders.values)
plt.title("Daily Orders Trend")
plt.xlabel("Date")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.show()
```

DAILY SALES

```python
daily_sales = (orders.groupby("order_date")["sales_amount"].sum())
```

```python
plt.figure(figsize=(12, 5))

plt.plot(daily_sales.index,daily_sales.values)

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales Amount")

plt.xticks(rotation=45)

plt.show()
```

MONTHLY SALES

```python
orders["year_month"] = (orders["order_date"].dt.to_period("M"))
```

```python
monthly_sales = (orders.groupby("year_month")["sales_amount"].sum())
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index.astype(str),monthly_sales.values)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")

plt.xticks(rotation=90)

plt.show()
```

CUSTOMER Vs ORDER ANALYSIS

```python
1 .ORDERS PER CUSTOMERS
2 .CUSTOMER SPENDING
3 .AVERAGE SPENDING PER CUSTOMER
4 .MERGE CUSTOMERS WITH ORDERS
5 .ADD CUSTOMER DEMOGRAPHICS
```

ORDERS PER CUSTOMERS

```python
orders_per_customer = (orders.groupby("user_id").size().sort_values(ascending=False))

orders_per_customer.head(10)
```

CUSTOMER SPENDING

```python
customer_spending = (orders.groupby("user_id")["sales_amount"].sum().sort_values(ascending=False))
customer_spending.head(10)
```

AVERAGE SPENDING PER CUSTOMER

```python
average_customer_spending = customer_spending.mean()

print("Average customer spending: ₹",round(average_customer_spending, 2))
```

MERGE CUSTOMERS WITH ORDERS

```python
customer_summary = (orders.groupby("user_id").agg(total_orders=("user_id", "size"),total_spend=("sales_amount", "sum"),
        total_quantity=("sales_qty", "sum")).reset_index())
```

```python
customer_summary.head()
```

CUSTOMER DEMOGRAPHICS

```python
customer_summary = customer_summary.merge(users,on="user_id",how="left")
```

```python
customer_summary.head()
```

```python
occupation_spending = (customer_summary.groupby("occupation")["total_spend"].mean().sort_values(ascending=False))

occupation_spending
```

```python
gender_spending = (customer_summary.groupby("gender")["total_spend"].mean().sort_values(ascending=False))

gender_spending
```
