Create the combined customer analytics dataset

```python

import os
import pandas as pd

data_path = r"C:\Users\ASUS\swiggy_customer_analytics\data\processed"

rfm = pd.read_csv(os.path.join(data_path, "customer_rfm.csv"))
users = pd.read_csv(os.path.join(data_path, "users_cleaned.csv"))
orders = pd.read_csv(os.path.join(data_path, "orders_cleaned.csv"))
restaurants = pd.read_csv(os.path.join(data_path, "restaurants_cleaned.csv"))

# Customer-level dataset: RFM contains one row per customer
customer_data = rfm.merge(
    users,
    on="user_id",
    how="left",
    suffixes=("_rfm", "_users")
)

print("Customer dataset shape:", customer_data.shape)
print("Unique customer IDs:", customer_data["user_id"].nunique())
print("\nFirst five rows:")
print(customer_data.head())
```

Calculate the main dashboard KPIs

```python

# Customer-level KPIs from the RFM dataset
total_customers = rfm["user_id"].nunique()
total_orders = rfm["total_orders"].sum()
total_spending = rfm["total_spend"].sum()
average_customer_spending = rfm["total_spend"].mean()
average_orders_per_customer = rfm["total_orders"].mean()

# Order-level KPIs from the orders dataset
order_revenue = orders["sales_amount"].sum()
average_order_sales = orders["sales_amount"].mean()

print("===== SWIGGY DASHBOARD KPIs =====")
print("Total customers:", f"{total_customers:,}")
print("Total orders (sum of customer totals):", f"{total_orders:,.0f}")
print("Total customer spending:", f"{total_spending:,.2f}")
print("Average spending per customer:", f"{average_customer_spending:,.2f}")
print("Average orders per customer:", f"{average_orders_per_customer:.2f}")
print("Sum of order sales amounts:", f"{order_revenue:,.2f}")
print("Average sales amount per order row:", f"{average_order_sales:,.2f}")
print("Number of order rows:", f"{len(orders):,}")
```

Create customer activity and risk labels

```python

# Create a copy so the original RFM data remains unchanged
customer_data = rfm.copy()

def get_customer_status(row):
    if row["frequency"] == 0:
        return "New Customer"
    elif row["recency"] <= 30:
        return "Active"
    elif row["recency"] <= 60:
        return "Cooling"
    elif row["recency"] <= 90:
        return "At Risk"
    else:
        return "Inactive"

customer_data["customer_status"] = customer_data.apply(
    get_customer_status, axis=1
)

def get_risk_level(row):
    if row["frequency"] == 0:
        return "Not Yet Rated"
    elif row["recency"] <= 30:
        return "Low Risk"
    elif row["recency"] <= 60:
        return "Medium Risk"
    else:
        return "High Risk"

customer_data["risk_level"] = customer_data.apply(
    get_risk_level, axis=1
)

print("Customer status distribution:")
print(customer_data["customer_status"].value_counts())

print("\nRisk level distribution:")
print(customer_data["risk_level"].value_counts())
```

Save the customer dataset for your web app

```python
import os

output_path = os.path.join(
    data_path,
    "customer_analytics_ready.csv"
)

customer_data.to_csv(output_path, index=False)

print("Saved successfully!")
print("File location:", output_path)
print("Rows:", len(customer_data))
print("Columns:", len(customer_data.columns))
```
